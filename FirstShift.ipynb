"FirstShift predicts whether a job posting is fake from its text and details, to help teens and young adults avoid job scams."

```python
!pip install -q kagglehub
import kagglehub, os
import pandas as pd
import numpy as np

path = kagglehub.dataset_download("shivamb/real-or-fake-fake-jobposting-prediction")
df = pd.read_csv(os.path.join(path, "fake_job_postings.csv"))
print(df.shape)
print(df["fraudulent"].value_counts())
```

```python
df.columns
```

```python
df.isna().sum()
```

```python
df[df["fraudulent"] == 1].head(20)
```

```python
for col in ["has_company_logo", "telecommuting", "required_experience", "employment_type"]:
    print(df.groupby(col)["fraudulent"].mean().round(3), "\n")
```

Observations: Fraudlent job listings tend to advertise telecomuting flexibility, as well as tending not to have company logos.
